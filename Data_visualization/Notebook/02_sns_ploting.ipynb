```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

### Step 1

```python
df=sns.load_dataset('penguins')
```

```python
df.head
```

```python
df.info
```

### Step 2

- Plot ki Type: Scatter Plot (Kyunki hum 2 numbers ka rishta dekhna chahte hain)

- X-axis (Neeche): 'bill_length_mm' (Choanch ki lambai)

- Y-axis (Oopar): 'body_mass_g' (Penguin ka wazan)

```python
sns.scatterplot(data=df,x='bill_length_mm', y='body_mass_g')
```

- Step 5 (Grouping): hue='species' (Penguins ki types ke hisab se alag color do)

- Step 6 (Size of plot): figsize=(10, 5) (Graph ki width aur height set karo)

- Step 7 (Colour of plot): palette='Set2' (Acche colors select karo)

- Step 8 (Style of plot): sns.set_theme(style='whitegrid') (Background clear karo)

```python
sns.scatterplot(data=df,x='bill_length_mm', y='body_mass_g' , hue='species' , palette='Set2')
plt.figure(figsize=(10,6))
sns.set_theme(style='whitegrid')
plt.title('Bill Length vs Body Mass by Species')
plt.xlabel('Bill Length (mm)')
plt.ylabel('Body Mass (g)')
plt.show()
```

### Interpreter

- Gentoo (Blue dots): Ye sab se bhari aur bade penguins hain! Inka Body Mass (weight) 4500g se 6000g+ tak hai aur inki choanch (bill) bhi lambi hai.

- Chinstrap (Orange dots): Inka weight chota/medium hai (3000g se 4500g), lekin inki choanch (bill length) kafi lambi hai (45mm se 55mm).

- Adelie (Green dots): Ye sab se chote penguins hain! Inka wazan bhi kam hai aur bill length sab se choti (35mm se 42mm) hai.

```python
sns.scatterplot(
    data=df, 
    x='bill_length_mm', 
    y='body_mass_g', 
    hue='species', 
    size='flipper_length_mm',  # Jo bada penguin hoga uska dot bhi bada dikhega
    sizes=(20, 200)
)
```

```python
sns.scatterplot(
    data=df, 
    x='bill_length_mm', 
    y='body_mass_g', 
    hue='species', 
    style='species'  # Adelie circle, Gentoo square, Chinstrap cross ban jayega
)
```

```python
sns.lmplot(
    data=df, 
    x='bill_length_mm', 
    y='body_mass_g', 
    hue='species'  # Har species ki apni trend line ban jayegi
)
```
