We need the answer to this question → **Category-wise Sales**

### Sales vs Profit

**Question:** Does higher Sales generally lead to higher Profit?

**Visualization:** Scatter Plot

**Reason:** A scatter plot is used to understand the relationship between two numerical variables, Sales and Profit.

**Conclusion:** The scatter plot shows a generally positive relationship between Sales and Profit. However, some high-Sales orders still result in a loss, showing that higher Sales do not always guarantee higher Profit.

```python
# scatter plot of discount vs profit
plt.scatter(df['Discount'],df['Profit'])
```

```python
# how many orders contains loss
df['Profit'].lt(0).sum()
```

```python
# how many orders contains profit
df['Profit'].ge(0).sum()
```

**Additional Insight:** Out of 51,290 orders, 38,746 orders generated profit, while 12,544 orders resulted in a loss.

```python
# scatter plot of quantity vs sales
plt.scatter(df['Quantity'],df['Sales'])
```

```python
plt.scatter(df['Quantity'], df['Sales'])

plt.title('Quantity vs Sales')
plt.xlabel('Quantity')
plt.ylabel('Sales')
plt.show()
```

- Overall, Quantity does not appear to have a strong linear relationship with Sales. Although higher quantities can sometimes result in higher Sales, the relationship is inconsistent and highly scattered.

## Next step is "seaborn se Data ko Beautiful & Dynamic banana".

```python
# Clean theme set karo
sns.set_theme(style="whitegrid")

# Category-wise Sales plot karna
plt.figure(figsize=(8, 4))
sns.barplot(data=df, x='Category', y='Sales', estimator=sum, errorbar=None, palette='viridis')

plt.title('Total Sales by Category', fontsize=12, fontweight='bold')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.show()
```

#### Step 5: Advanced Scatter Plot (Discount vs Profit by Category)

```python
plt.figure(figsize=(8, 5))

# Category ke hisab se colors shift honge (hue)
sns.scatterplot(data=df, x='Discount', y='Profit', hue='Category', s=70)

plt.title('Impact of Discount on Profit by Category', fontsize=12, fontweight='bold')
plt.axhline(0, color='red', linestyle='--') # 0 line par loss marker red line
plt.show()
```

### Step 5 Conclusion: Impact of Discount on Profit
* **Discount Threshold:** Discounts up to 20% preserve positive profit margins. Any discount above 20% directly pushes transactions into net loss territory (below the red zero-profit baseline).
* **High-Risk Categories:** Heavy discounting (>50%) on Technology and Office Supplies causes extreme financial losses, reaching down to -$6,000 on single orders.
* **Actionable Recommendation:** Implement a strict capping rule on promotional discounts (maximum limit of 15%-20%) across all product categories to prevent revenue erosion.

## Step 6: Final 2x2 EDA Dashboard Screen
Combined key categorical, relational, and temporal visual metrics into a consolidated single-screen layout using Matplotlib Subplots.


```python
# Clean grid style
sns.set_theme(style="whitegrid")

# 2x2 grid ka layout banana
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Sales & Profit Analysis Dashboard', fontsize=16, fontweight='bold')

# Chart 1: Category Sales
sns.barplot(ax=axes[0, 0], data=df, x='Category', y='Sales', estimator=sum, errorbar=None, palette='viridis')
axes[0, 0].set_title('Total Sales by Category')

# Chart 2: Category Profit
sns.barplot(ax=axes[0, 1], data=df, x='Category', y='Profit', estimator=sum, errorbar=None, palette='magma')
axes[0, 1].set_title('Total Profit by Category')

# Chart 3: Discount vs Profit Scatter
sns.scatterplot(ax=axes[1, 0], data=df, x='Discount', y='Profit', hue='Category')
axes[1, 0].axhline(0, color='red', linestyle='--')
axes[1, 0].set_title('Discount vs Profit Impact')

# Chart 4: Yearly Profit Trend Line
yearly_profit = df.groupby(df['Order Date'].dt.year)['Profit'].sum()
axes[1, 1].plot(yearly_profit.index, yearly_profit.values, marker='o', color='teal', linewidth=2)
axes[1, 1].set_title('Yearly Profit Trend')
axes[1, 1].set_xlabel('Year')
axes[1, 1].set_ylabel('Profit')

plt.tight_layout()
plt.show()
```
