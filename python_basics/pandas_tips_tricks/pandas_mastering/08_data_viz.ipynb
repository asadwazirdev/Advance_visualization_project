### Data visualization

```python
# import all the libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ydata_profiling as yp



```

```python
# load the dataset 
df = sns.load_dataset("titanic")
```

```python
# to make the plot of the dataset
df.plot(kind = 'scatter', x = 'age', y = 'fare', color = 'black')
```

```python
#to make the selected coloumn`s plot
df.groupby('sex')['fare'].mean().plot(kind = 'bar', color = 'blue')
```

```python
# Bar plot using seaborn on df
sns.barplot(data = df, x = 'sex', y = 'fare', hue = 'class')
```
